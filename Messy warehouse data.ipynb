```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\asus\\OneDrive\\Desktop\\project\\warehouse_messy_data.csv")
```

```python
df.head(5)
```

```python
# Strip leading and trailing spaces from string columns 
for column in df.select_dtypes(include = ['object']).columns: df[column]=df[column].str.strip()
```

```python
df.head(-10)
```

```python
# Convert Product Name to proper case 
df['Product Name']= df['Product Name'].str.title()
```

```python
# convert category
df['Category']= df['Category'].str.capitalize()
```

```python
df.head(5)
```

```python
# Correct the Quantity column 
df['Quantity']= df['Quantity'].replace('two hundred',200) # replaces 'two hundred' with 200
# replace NAN with np.nan
df['Quantity']= df['Quantity'].replace('NAN', np.nan)
#convert to numeric
df['Quantity']= pd.to_numeric(df['Quantity'], errors ='coerce')
```

```python
# correcting the price column 
df['Price']= df['Price'].replace('NAN',np.nan)
df['Price']= pd.to_numeric(df['Price'], errors = 'coerce')
```

```python
# correcting the last restocked column
df['Last Restocked'] = df['Last Restocked'].replace('NAN', np.nan)
df['Last Restocked'] = pd.to_datetime(df['Last Restocked'], errors ='coerce')
```

```python
df.head(5)
```

*Handling Missing Values*

```python
# Fill NaN values with a default value or strategy, here we use mean for numeric columns
df['Quantity'].fillna(df['Quantity'].mean(), inplace = True)
df['Price'].fillna(df['Price'].mean(), inplace = True)
```

```python
# Fill NaN values for categorical columns with a placeholder or most frequent value 
df['Product Name'].fillna('Unknown Product', inplace=True) 
df['Category'].fillna('Unknown Category', inplace=True) 
df['Warehouse'].fillna(df['Warehouse'].mode()[0], inplace=True) 
df['Location'].fillna('Unknown Location', inplace=True) 
df['Supplier'].fillna(df['Supplier'].mode()[0], inplace=True) 
df['Status'].fillna(df['Status'].mode()[0], inplace=True) 
df['Last Restocked'].fillna(pd.to_datetime('today'), inplace=True)
```

```python
df.head(5)
```

```python
# Save the cleaned data to a new CSV file 
df.to_csv('cleaned_warehouse_messy_data.csv', index=False)
```
